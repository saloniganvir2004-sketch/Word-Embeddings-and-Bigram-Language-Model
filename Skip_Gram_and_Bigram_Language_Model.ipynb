```python
import numpy as np

# Training sentence
sentence = "i love machine learning and i love deep learning"

# Create vocabulary
words = sentence.split()
vocab = list(set(words))

word_to_index = {word: i for i, word in enumerate(vocab)}
index_to_word = {i: word for word, i in word_to_index.items()}

vocab_size = len(vocab)
embedding_dim = 5
learning_rate = 0.01
epochs = 1000

# Generate Skip-gram training pairs
window_size = 2
training_pairs = []

for i, word in enumerate(words):
    for j in range(max(0, i - window_size), min(len(words), i + window_size + 1)):
        if i != j:
            training_pairs.append(
                (word_to_index[word], word_to_index[words[j]])
            )

# Initialize input and output weights
np.random.seed(42)

W = np.random.randn(vocab_size, embedding_dim) * 0.01
W_out = np.random.randn(embedding_dim, vocab_size) * 0.01


# Softmax function
def softmax(x):
    exp_x = np.exp(x - np.max(x))
    return exp_x / np.sum(exp_x)


# Training using Gradient Descent
for epoch in range(epochs):

    total_loss = 0

    for input_word, target_word in training_pairs:

        # Forward pass
        h = W[input_word]

        u = np.dot(h, W_out)

        y_pred = softmax(u)

        # One-hot target
        y_true = np.zeros(vocab_size)
        y_true[target_word] = 1

        # Cross-entropy loss
        loss = -np.log(y_pred[target_word] + 1e-10)
        total_loss += loss

        # Gradient
        error = y_pred - y_true

        grad_W_out = np.outer(h, error)
        grad_h = np.dot(W_out, error)

        # Gradient descent
        W_out -= learning_rate * grad_W_out
        W[input_word] -= learning_rate * grad_h

    if (epoch + 1) % 100 == 0:
        print("Epoch:", epoch + 1, "Loss:", total_loss)


# Display word embeddings
print("\nWord Embeddings:")
for word in vocab:
    print(word, ":", W[word_to_index[word]])

# Find embedding of a particular word
word = "learning"

print("\nEmbedding of", word, ":")
print(W[word_to_index[word]])
```

```python
from collections import defaultdict, Counter

# Training text
text = "i love machine learning i love deep learning machine learning is interesting"

# Convert text into words
words = text.lower().split()

# Bigram data structure
bigram_model = defaultdict(Counter)

# Build bigram model
for i in range(len(words) - 1):
    current_word = words[i]
    next_word = words[i + 1]

    bigram_model[current_word][next_word] += 1


# Display bigram model
print("BIGRAM MODEL:\n")

for word, next_words in bigram_model.items():
    print(word, "->", dict(next_words))


# Prediction function
def predict_next_word(word):
    if word not in bigram_model:
        return "Word not found"

    next_words = bigram_model[word]

    # Select the most frequently occurring next word
    predicted_word = next_words.most_common(1)[0][0]

    return predicted_word


# Take user input
word = input("\nEnter a word: ").lower()

prediction = predict_next_word(word)

print("Predicted next word:", prediction)
```
