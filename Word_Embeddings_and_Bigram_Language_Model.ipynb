```python
import numpy as np

text = "i love machine learning i love deep learning"
words = text.split()
vocab = list(set(words))
word_to_id = {w: i for i, w in enumerate(vocab)}
V = len(vocab)
D = 10
lr = 0.01
epochs = 100

pairs = []
for i, word in enumerate(words):
    for j in range(max(0, i-2), min(len(words), i+3)):
        if i != j:
            pairs.append((word_to_id[word], word_to_id[words[j]]))

W1 = np.random.randn(V, D) * 0.01
W2 = np.random.randn(D, V) * 0.01

for epoch in range(epochs):
    loss = 0
    for center, context in pairs:
        h = W1[center]
        scores = h @ W2
        scores -= np.max(scores)
        probs = np.exp(scores) / np.sum(np.exp(scores))
        loss -= np.log(probs[context] + 1e-10)

        ds = probs.copy()
        ds[context] -= 1

        dW2 = np.outer(h, ds)
        dW1 = W2 @ ds

        W2 -= lr * dW2
        W1[center] -= lr * dW1

    if (epoch + 1) % 20 == 0:
        print(f"Epoch {epoch+1}, Loss: {loss:.4f}")

print("\nWord Embeddings:")
for word in vocab:
    print(word, W1[word_to_id[word]])
```

```python
from collections import Counter, defaultdict

text = "i love machine learning i love deep learning"
words = text.lower().split()

bigram_count = Counter()
word_count = Counter()
bigram_model = defaultdict(dict)

for i in range(len(words) - 1):
    w1, w2 = words[i], words[i + 1]
    bigram_count[(w1, w2)] += 1
    word_count[w1] += 1

for (w1, w2), count in bigram_count.items():
    bigram_model[w1][w2] = count / word_count[w1]

print("Bigram Probabilities:")
for w1 in bigram_model:
    print(w1, bigram_model[w1])

word = input("\nEnter a word: ").lower()

if word in bigram_model:
    next_word = max(bigram_model[word], key=bigram_model[word].get)
    probability = bigram_model[word][next_word]
    print("Predicted Next Word:", next_word)
    print("Probability:", round(probability, 4))
else:
    print("Word not found in the model.")
```
